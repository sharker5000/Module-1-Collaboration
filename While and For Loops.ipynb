```python
secret = 7
guess = 3
if guess < secret:
    print("Too low")
    guess = int(input("guess a number between 1 and 10: "))
elif guess > secret:
    print("Too high")
    guess = int(input("guess a number between 1 and 10: "))
elif guess == secret:
    print("Just right!")
```

```python
small = True
green = False
if (small == True) and (green == True):
    print("It's a pea")
elif (small == True) and (green == False):
    print("It's a cherry")
elif (small == False) and (green == True):
    print("It's a watermelon")
elif (small == False) and (green == False):
    print("It's a pumpkin")
```

```python
guess_me = 7
number = 1
while number >= 0:
    if number < guess_me:
        print(number)
        print("too low")
        number += 1
    if number == guess_me:
        print(number)
        print("found it!")
        break
    if number > guess_me:
        print("oops!")
        break
```

```python
guess_me = 5
for number in range(10):
    if number < guess_me:
        print(number)
        print("too low")
    if number == guess_me:
        print(number)
        print("found it!")
        break
    if number > guess_me:
        print("oops!")
        break
```
