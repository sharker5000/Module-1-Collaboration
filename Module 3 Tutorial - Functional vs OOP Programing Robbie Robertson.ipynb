```python
class Solution:
    def sort012(self, arr):
        zeros = []
        ones = []
        twos = []
        for i in arr:
            if i == 0:
                zeros.append(i)
            elif i == 1:
                ones.append(i)
            elif i == 2:
                twos.append(i)
        x = (zeros + ones + twos)
        for i in x:
            arr.pop(0)
            arr.append(i)
```

```python
class Solution:
    def firstSearch(self, arr, k):
        found = False
        for i in arr:
            if i == k:
                return(arr.index(i))
                found = True
        
        if found == False:
            return(-1)
```
