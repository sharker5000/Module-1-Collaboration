```python
things = ["mozzarella", "cinderella", "salmonella"]
print(things)
```

```python
things = ["mozzarella", "cinderella", "salmonella"]
x = things.index("cinderella")
things[x] = things[x].capitalize()
print(things)

# no, it did not change the element in the list. It needed to be changed by having things[x] = things[x].capitalize()
```

```python
things = ["mozzarella", "cinderella", "salmonella"]
x = things.index("mozzarella")
things[x] = things[x].upper()
print(things)
```

```python
things = ["mozzarella", "cinderella", "salmonella"]
things.remove("salmonella")
print(things)
```

```python
def get_odds(x):
    odds = []
    for i in range(x):
        if i % 2 != 0:
            odds.append(i)
    print(odds[3])
get_odds(10)
```
