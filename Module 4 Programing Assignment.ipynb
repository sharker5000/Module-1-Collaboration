```python
import sqlite3

conn = sqlite3.connect("books.db")
lib = conn.cursor()
lib.execute('''CREATE TABLE books
    (title VARCHAR(80),
    author VARCHAR(80),
    year INT)''')

conn.close()
lib.close()
```

```python
import sqlite3
import csv

conn = sqlite3.connect("books.db")
ins = 'INSERT INTO books (title, author, year) VALUES(?, ?, ?)'
lib = conn.cursor()
with open('books2.csv','rt') as books2:
    reader = csv.reader(books2)
    for row in reader:
        lib.execute(ins, row)

conn.close()
```

```python
import sqlalchemy as sa

engine = sa.create_engine('sqlite:///books.db')
with engine.connect() as connection:
    query = sa.text('SELECT title FROM books ORDER BY title ASC')
    rows = connection.execute(query)
    for row in rows:
        print(row.title)
```
